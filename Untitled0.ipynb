```python
import torch
import evaluate
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, TrainingArguments, Trainer, DataCollatorForSeq2Seq
from datasets import load_dataset

# Check if GPU (T4) is available
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Load CNN/DailyMail dataset
dataset = load_dataset("cnn_dailymail", "3.0.0")

# Use a smaller dataset for faster training
dataset["train"] = dataset["train"].shuffle(seed=42).select(range(10000))  # 10K samples for training
dataset["validation"] = dataset["validation"].shuffle(seed=42).select(range(2000))  # 2K for validation
```

```python
model_name = "t5-small"  # Can switch to "google/pegasus-cnn_dailymail" for better results
tokenizer = AutoTokenizer.from_pretrained(model_name)

def preprocess_function(examples):
    inputs = ["summarize: " + text for text in examples["article"]]
    model_inputs = tokenizer(inputs, max_length=512, truncation=True, padding="max_length")

    with tokenizer.as_target_tokenizer():
        labels = tokenizer(examples["highlights"], max_length=128, truncation=True, padding="max_length")

    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

tokenized_dataset = dataset.map(preprocess_function, batched=True)
```

```python
model = AutoModelForSeq2SeqLM.from_pretrained(model_name).to(device)

data_collator = DataCollatorForSeq2Seq(tokenizer, model=model)



training_args = TrainingArguments(
    output_dir="./results",
    evaluation_strategy="epoch",
    learning_rate=5e-5,
    per_device_train_batch_size=1,  # Prevents OOM errors
    per_device_eval_batch_size=1,
    num_train_epochs=1,  # Train for 1 epoch instead of 3
    weight_decay=0.01,
    fp16=True,  # Mixed precision for speed
    save_total_limit=1,
    logging_dir="./logs",
    logging_steps=1000,  # Log less frequently to reduce overhead
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["validation"],
    data_collator=data_collator,
)


trainer.train()
```

```python
def summarize(text):
    inputs = tokenizer("summarize: " + text, return_tensors="pt", max_length=512, truncation=True).to(device)
    summary_ids = model.generate(inputs["input_ids"], max_length=150, min_length=50, length_penalty=2.0, num_beams=4)
    return tokenizer.decode(summary_ids[0], skip_special_tokens=True)

"""# Test on a sample article
sample_text = dataset["test"][0]["article"]
print("Original:", sample_text)
print("Summary:", summarize(sample_text))"""

# Test on a sample article
sample_text = "A leading technology company has unveiled a groundbreaking advancement in artificial intelligence, claiming it will revolutionize industries ranging from healthcare to finance. The new AI model, designed to process and analyze massive datasets at unprecedented speeds, is expected to improve medical diagnoses, automate complex tasks, and enhance customer interactions. Experts believe this innovation could significantly impact the global job market, sparking debates on the future of AI regulations and ethical considerations. The company plans to roll out the technology in the coming months, promising a new era of efficiency and automation."
print("Original:", sample_text)
print("Summary:", summarize(sample_text))
```

```python
rouge = evaluate.load("rouge")

def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    decoded_preds = tokenizer.batch_decode(predictions, skip_special_tokens=True)
    decoded_labels = tokenizer.batch_decode(labels, skip_special_tokens=True)
    return rouge.compute(predictions=decoded_preds, references=decoded_labels)

results = trainer.evaluate()
print(results)
```
